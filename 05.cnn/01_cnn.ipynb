```python
c1 = nn.Conv2d(1, 8, 6, stride=2)
x = torch.randn(32, 1, 28, 28)
# print(c1(x))
c2 = nn.Conv2d(8, 16, 3, padding=1)
# print(c2(c1(x)).shape)
mp = nn.MaxPool2d(kernel_size=2, stride=2)
print(mp(c2(c1(x))))
```

```python
mp = nn.MaxPool2d(2)  # kernel_size=2, stride=2와 동일
x = torch.randn(1, 6, 6)
print(x)
print(mp(x))
print(mp(torch.randn(32, 3, 6, 6)).shape)
```

```python
class CNN(nn.Module):
    def __init__(self):
        super().__init__()
        self.c1 = nn.Conv2d(1, 8, 6, stride=2)
        self.c2 = nn.Conv2d(8, 16, 3, padding=1)
        self.mp = nn.MaxPool2d(2)
        self.fc = nn.Linear(16 * 6 * 6, 10)

    def forward(self, x):
        x = self.c1(x)
        x = self.c2(x)
        x = self.mp(x)
        x = torch.flatten(x, start_dim=1)
        x = self.fc(x)
        return x


x = torch.randn(32, 1, 28, 28)
model = CNN()
print(model(x).shape)
```
