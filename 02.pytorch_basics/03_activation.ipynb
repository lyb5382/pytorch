### activation function
: 신경망의 선형 연산 결과에 비선형성을 추가하여 복잡한 패턴을 학습할 수 있도록 돕는 함수
- 활성화 함수의 목적
    - 은닉층에 비선형성 추가
    - 출력값을 조절
    - 문제에 맞는 출력 생성
#### sigmoid 함수
: 입력값 0~1사이의 값으로 변환하는 함수
- 특징
    - 출력 범위가 0과 1사이
    - 입력값이 0이면 출력값은 0.5
    - 입력값이 커지면 출력은 1에 가까워짐
    - 입력값이 작아지면 출력은 0에 가까워짐
    - 이진분류에서 한 클래스에 속할 확률처럼 해석할 수 있는 값들을 만들때 사용
- 단점
    - 입력값의 절대값이 아주 커지면 기울기 작아지는 포화현상이 나타남
    - 은닉층에 여러번 사용하면 기울기 소실 문제가 생길 수 있음

```python
import torch
import torch.nn as nn
```

```python
x = torch.tensor([-10.0, -2, -1, 0, 1, 2, 10])
op = torch.sigmoid(x)
print(op)
```

```python
logits = torch.tensor([-2, -0.5, 0, 0.5, 2])
print("모델을 통과한 출력(logits):", logits)
si = torch.sigmoid(logits)
print("sigmoid 출력:", si)
c0 = (si >= 0.5).int()  # .int는 불리안을 정수형으로 변환
```

#### ReLu 함수
: 입력값이 0보다 작으면 0 출력, 0보다 크면 입력값 그대로 출력
- 특징
    - 계산이 단순하고 효율적임
    - 양수 영역에서는 기울기가 1이므로 기울기 소실 문제 완화
    - 일반적인 신경망의 은닉층에서 널리 사용됨
- 단점
    - 음수 영역에서는 출력과 기울기가 0임
    - 학습 중 일부 뉴런이 음수 영역에 머물면 업데이트가 잘 되지 않음

```python
x = torch.tensor([-5.0, -2, -1, 0, 1, 2, 5])
op=torch.relu(x)
print('입력:',x)
print('출력:',op)
```

#### tanh
: 입력값이 -1~1사이의 값으로 변환
- 특징
    - 입력값이 0이면 출력값도 0
    - 양수입력은 1에 가깝고 음수입력은 -1에 가까움
    - 출력중심이 0
- 단점
    - 입력값의 절대값이 매우 커지면 기울기가 작아지는 포화현상이 발생

```python
x = torch.tensor([-10.0, -2, -1, 0, 1, 2, 10])
op = torch.tanh(x)
print(x, op)
```

#### softmax
: 입력받은 실수 값들은 0과 1사이의 값으로 변환하고, 이 값들의 총합이 정확히 1이 되도록 만들어 확률분포로 해석할 수 있게 해줌<br>
주로 서로 베타적인 여러 클래수 중 하나를 선택하는 다중분류의 출력에 사용됨
- 특징
    - 각 출력값은 0~1사이의 값
    - 모든 클래스의 출력값을 합치면 1이 됨
    - 출력값이 각 클래스에 대한 예측 확률처럼 해석할 수 있음
    - 가장 큰 출력값을 가진 클래스가 모델의 예측 클래스가 됨
- 옵션 dim
    - dim=0: 첫번째 차원을 기준으로 softmax계산
    - dim=1: 두번째 차원을 기준으로 softmax계산
    - batch데이터가 (batch_size, num_class) 형태라면 일반적으로 dim=1을 사용함

```python
logits = torch.tensor([2, 1, 0.1])
sm = torch.softmax(logits, dim=0)
print("logits:", logits)
print("softmax:", sm)
print("확률의 합:", sm.sum())
print("예측 클래스 인덱스:", torch.argmax(sm))
print(torch.argmax(logits))
```

- 예를 들어 데이터 2개를 각각 3개 클래스로 분류한다면 shape(2, 3)
    - 각 행(row)은 하나의 데이터에 대한 클래스의 합이 1이 되어야 함
    - 각 데이터별로 클래스 확률의 합이 1이 되어야 함
- 비교
    - dim=0: 같은 클래스 열을 기준으로 서로 다른 데이터 사이의 합이 1이 됨
    - dim=1: 각 데이터의 클래스 확률합이 1이 됨

```python
logits = torch.tensor([[2.0, 1.0, 0.1], [0.5, 2.5, 1.0]])
d0 = torch.softmax(logits, dim=0)
d1 = torch.softmax(logits, dim=1)
print("dim0의 결과, dim0의 열방향의 합", d0, d0.sum(dim=0))
print("dim1의 결과, dim1의 행방향의 합", d1, d1.sum(dim=1))
```

#### 4가지 활성화 함수
|함수|출력범위|대표적인 사용 위치|
|:---:|:---:|:---:|
|sigmoid|0~1|이진분류 출력층|
|ReLU|0~양의 무한대|일반적인 은닉층|
|tanh|-1~1|일부 은닉층|
|softmax|0~1 전체합=1|다중 분류 출력층|

```python
x = torch.tensor([-2.0, -1, 0, 1, 2])
print(torch.sigmoid(x), torch.relu(x), torch.tanh(x))
cs = torch.tensor([2.0, 1, 0.1])
print(torch.softmax(cs, dim=0))
```
