### pytorch 주요 수학 함수
#### 1. 기본 수학 함수
|함수|의미|예시|설명|
|:---:|:---:|:---:|:---:|
|torch.abs(a)|절댓값|abs(-3)=3|모든 원소의 절대값 반환|
|torch.sqrt(a)|제곱근|sqrt(9)=3|$\sqrt{x}$|
|torch.square(a)|제곱|square(3)=9|$x^2$|
|torch.pow(a, n)|거듭제곱|pow(2, 3)=8|$x^n$|
|torch.exp(a)|지수함수|exp(1)=e|$e^x$|
|torch.log(a)|자연로그|log(e)=1|ln(x)|
|torch.log10(a)|상용로그|log10(10)=1|$\log_{10}(x)$|
|torch.log2(a)|밑이 2인 로그|log2(2)=1|$\log_2(x)$|

```python
a = torch.randn(3, 3)
print(a)
print(torch.abs(a))
print(torch.sqrt(torch.abs(a)))
print(torch.square(a))
print(torch.pow(a, 2))

print(torch.exp(a))
print(torch.log(torch.abs(a)))
print(torch.log(torch.exp(torch.tensor(1))))
print(torch.log10(torch.tensor(10)))
```

|함수|의미|예시|
|:---:|:---:|:---:|
|torch.min(a)|최솟값|전체 원소 중 가장 작은 값|

#### torch.nan
- not a number: 수학적으로 정의할 수 없거나 컴퓨터가 정상적인 실수 값으로 표현할 수 없는 연산 불능 / 수치적인 오류 상태 의미
    - 정의되지 않은 수학
        - 연산음수의 로그 계산: torch.log(torch.tensor(-1.0)) => nan
        - 0을 0으로 나누기
        - 무한대 간의 뺄샘: inf-inf => nan
    - 딥러닝 학습 중 수치적 불안정성(gradient explotion 등)
        - 그라디언트 폭주: 가중치 업데이트량이 너무 커져 수치가 표현 범위 초과 시
        - 언더플로우/오버플로우

```python
torch.nan
# print(torch.log(torch.tansor(-1)))
print(torch.isnan(torch.tensor([1, 2, torch.nan, 3, 4])))
print(torch.isinf(torch.tensor([1, 2, 3, 4, torch.inf])))
a=torch.randn(6,1)
print(a)
```

```python
print(torch.sort(a))
print(torch.sort(a, dim=0))
print(a.sort(dim=0))
print(a.sort(dim=0, descending=True))
print(torch.max(a))
print(a.max())
print(torch.abs(a))
print(a.abs())
```

#### reshape()
: 텐서의 원소 개수는 유지한 채 모양(shape)만 변경할 때 사용

```python
a = torch.randint(1, 5, size=(12,))
print(a)
print(a.shape)
b = a.reshape(2, 2, 3)
print(b)
print(b.ndim)
```

```python
a = torch.arange(
    20,
)
print(a)
print(a.reshape(4, 5))
print(a.reshape(2, -1))  # -1은 알아서 맞추라는 뜻
print(a.reshape(2, -1).shape)
print(a.reshape())
```

#### 기타 함수
##### transpose(): 2개의 차원 교환
- 주로 2차원 행렬의 전치(low<->column)나 특정 두 축의 위치만 바꿀때 사용
- 차원의 개수 상관 없이 정확히 2개 축의 인덱스만 넘겨줘야 함
##### permute(): 모든 차원의 순서를 재배치
- 3차원 이상의 고차원 텐서의 차원 순서를 한번에 뒤섞을 때 사용
##### t() / T: 행렬 연산(전치)을 간결하게 쓰기 위한 메서드 및 속성
- .t(): 2차원 텐서 전용 전치 함수
- .T: 모든 차원을 역순으로 뒤집은 결과 반환
    - 2차원 행렬: 일반적인 전치 행력(A^T)이 됨
    - 고차원 텐서: 차원의 순서가 완전히 반대로 뒤집힘 (2, 3, 4) -> (4, 3, 2)

```python
a = torch.tensor([1, 2, 3])
b = torch.tensor([2, 2, 1])
print(torch.sum(a * b))

a = a.reshape(3, 1)
b = b.reshape(3, 1)
print(a.transpose(1,0)@b)

```

```python
a = torch.ones(3, 4)
b = torch.zeros(3, 4)
c = torch.vstack([a, b])
d = torch.hstack([a, b])
e = torch.cat([a, b], dim=0)
f = torch.cat([a, b], dim=1)
print(c, d, e, f)
g = torch.ones(2, 3, 4)
h = torch.zeros(2, 3, 4)
print(torch.vstack([g, h]))
```

```python
a = torch.tensor([[1, 2], [3, 4]])
b = a
b[0, 0] = 100
print(a, b)
c = a.clone()
c[0, 1] = 200
print(a, c)
```
