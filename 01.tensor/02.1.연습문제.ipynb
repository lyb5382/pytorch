#### 문제 01

학급별 독서량을 나타내는 **3행 4열** 텐서 `books`를 직접 생성.

| 학급 | 1주 | 2주 | 3주 | 4주 |
|---|---:|---:|---:|---:|
| A | 8 | 12 | 10 | 15 |
| B | 11 | 9 | 14 | 13 |
| C | 7 | 16 | 12 | 10 |

**요구사항**
1. `books` 생성.
2. `dtype`, `shape`, `ndim`, `numel()`을 각각 출력.
3. 데이터 전체의 원소 개수 확인.

**제약:** `torch.tensor()`와 실습한 속성·메서드만 사용하기.


````python
import torch
````

````python
books = torch.tensor([[8, 12, 10, 15], [11, 9, 14, 13], [7, 16, 12, 10]])
print(books.dtype, books.shape, books.ndim, books.numel())
````

#### 문제 03

과일 매장의 판매 수량이 다음과 같음.

```python
sales = torch.tensor([[6, 2, 9],
                      [4, 8, 3]], dtype=torch.float32)
```

**요구사항**
1. `sales`와 동일한 Shape과 dtype의 0 텐서를 `returns`에 생성.
2. 반품 데이터에서 첫 번째 행의 두 번째 원소를 `1`, 두 번째 행의 세 번째 원소를 `2`로 변경.
3. `net = sales - returns`를 계산하고 `net`과 `net.dtype`을 출력.


````python
sales = torch.tensor([[6.0, 2.0, 9.0], [4.0, 8.0, 3.0]])
returns = torch.zeros_like(sales)
returns[0, 1] = 1
returns[1, 2] = 2
net = sales - returns
print(net, net.dtype)
````

#### 문제 04

제품 2종의 판매 수량과 단가가 다음과 같음.

```python
qty = torch.tensor([[2, 3],
                    [4, 1]])
price = torch.tensor([[100, 200],
                      [300, 400]])
```

**요구사항**
1. 같은 위치끼리 곱한 결과를 `elementwise`에 저장.
2. 행렬곱 결과를 `matrix_product`에 저장.
3. 두 결과를 출력하여 차이 확인.

**제약:** 원소별 곱은 `*`, 행렬곱은 `@` 사용하기.


````python
qty = torch.tensor([[2, 3], [4, 1]])
price = torch.tensor([[100, 200], [300, 400]])
elementwise = qty * price
matric_product = qty @ price
print(elementwise, matric_product)
````
