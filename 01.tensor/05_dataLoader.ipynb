```python
import torch
import pandas as pd
from torch.utils.data import DataLoader, TensorDataset
```

### 데이터 준비

```python
path = "https://raw.githubusercontent.com/DA4BAM/dataset/master/advertising.csv"
data = pd.read_csv(path)
print(data.head())
di = data[:10]
tg = "Sales"
x = di.drop(tg, axis=1)
y = di.loc[:, tg]
print(x.shape, y.shape)
```

#### 텐서로 변환

```python
print(type(x), type(y))
xn = x.values
yn = y.values
print(type(xn), type(yn))
print(xn.dtype, yn.dtype)
print(xn.shape, yn.shape)
yn = yn.reshape(10, 1)
```

```python
# 텐서로 변환
# dtype: float64 -> torch는 float32를 기본적인 가중치로 사용
xt = torch.tensor(xn, dtype=torch.float32)
yt = torch.tensor(yn, dtype=torch.float32)
print(xt.shape, yt.shape)
```

```python
xt[:5]
```

```python
yt[:5]
```

#### date loader 준비
- data loader는 텐서 데이터셋을 입력으로 받음
    - 텐서 x, y를 하나의 텐서 데이터셋으로 합친 후 dataloader생성
#### 텐서 데이터셋

```python
ds = TensorDataset(xt, yt)
ds[:5]
ds.tensors
ds.tensors[0]
len(ds)
```

#### 데이터 로더

```python
dl = DataLoader(ds, batch_size=3, shuffle=True)
```

- batch_size: 한번에 모델에 전달할 데이터 개수
- shuffle: 학습 데이터를 매 에포크 시작 시 무작위로 할지
- 에포크(epoch): 모델이 전체 학습 데이터를 모두 학습하는 한번
<br><br>
- 데이터로부터 데이터 추출하기

```python
for xb, yb in dl:
    print(xb, yb)
```

```python
for bi, (xb, yb) in enumerate(dl):
    print(bi, xb, yb)
```

문제 01
한 물류센터에서 상품별 입고량을 무작위로 생성하려고 함.

요구사항

torch.manual_seed(21)을 설정.
torch.rand()를 이용해 3행 4열의 0 이상 1 미만 실수 텐서 uniform_data 생성.
torch.randn()을 이용해 3행 4열의 표준정규분포 난수 텐서 normal_data 생성.
torch.randint()를 이용해 3행 4열의 20 이상 51 미만 정수 텐서 integer_data 생성.
세 텐서를 각각 출력.

```python
uniform_data = torch.rand(3, 4)
normal_data = torch.randn(3, 4)
integer_data = torch.randint(3, 4, (20, 51))
print(uniform_data, normal_data, integer_data)
```

```python
measurement = torch.tensor([1.0, 4.0, 9.0, 16.0])
sqrt_data = torch.sqrt(measurement)
square_dat = torch.square(measurement)
print(
    sqrt_data,
    square_dat,
    torch.exp(measurement),
    torch.log(measurement),
    torch.log2(measurement),
    torch.log10(measurement),
)
```
