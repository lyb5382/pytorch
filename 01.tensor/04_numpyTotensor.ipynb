### numpy 배열을 tensor로 변환
: 임베딩 벡터 처리, 데어터 전처리, 모델 입력 데이터 구성 등에서 많이 사용

```python
import numpy as np
import torch
```

- torch.from_numpy(array): 메모리 공유, numpy 배열을 tensor로 변환

```python
arr = np.array([1.0, 2, 3, 4])
x = torch.from_numpy(arr)
print(arr, x)
arr[0]=100
```

- torch.tensor(array): 데이터 복사, 원본과 분리된 tensor 생성

```python
arr = np.array([1.0, 2, 3, 4])
x = torch.tensor(arr)
arr[0] = 100
print(arr, x)
```

- torch.as_tensor(array): 가능하면 메모리 공유, 필요한 경우에만 데이터 복사
    - torch.from_numpy()와 비슷하게 동작하지만 입력 자료형이나 장치(device)등을 지정할 수 있음

```python
arr = np.array([1.0, 2, 3, 4])
x = torch.as_tensor(arr)
arr[0] = 100
print(arr, x)
```

```python
arr = np.array([1, 2, 3, 4], dtype=np.int32)
x = torch.as_tensor(arr, dtype=torch.float32)
```

#### pytorch tensor를 numpy 배열로 반환
: 모델이 계산한 결과를 python의 다른 라이브러리로 전달하거나, 데이터를 시각화하고 분석할때

```python
x = torch.tensor([0.1, 0.2, 0.3, 0.4])
arr = x.numpy()
print(x, arr, arr.dtype, arr.shape)
x[0] = 0.7
print(x, arr)
```

```python
x = torch.tensor([1.0, 2, 3], device="cuda")
arr = x.cpu().numpy()
```
