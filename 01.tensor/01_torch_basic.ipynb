### pytorch
: 인공지능과 딥러닝의 모델을 학습시키기 위한 파이썬 기반의 오픈소스 머신러닝 프레임워크
- 핵심 기능
    - tensor: 데이터를 다차원의 배열로 저장하고 계산
    - gpu 연산: gpu를 이용한 대량의 계산을 빠르게 처리
    - 자동미분(autograd): 학습에 필요한 기울기를 자동 계산
    - 최적화(qptimizer): 모델의 가중치를 수정하며 학습

##### ◈ 초기화 되지 않은 텐서

```python
x = torch.empty(4, 2)
x
```

##### ◈ 0 또는 1로 채워진 텐서

```python
x = torch.zeros(4, 2)
print(x, x.dtype)
y = torch.ones(2, 4)
print(y, y.dtype)
```

##### ◈ 사용자가...못봄ㅠ

```python
a = torch.tensor([1, 2, 3, 4])
print(a, type(a), a.dtype, a.shape)
b=torch.tensor([1,2,3.1,4])
print(b, type(b), b.dtype, b.shape)
```

#### ◈ 기초함수

```python
c = torch.tensor([[1, 2, 3], [4, 5, 6]])
# 행렬이라 각 행의 열의 개수가 같아야 함 ↓오류
# d = torch.tensor([[1, 2], [3, 4, 5]])
print(c, c.shape, c.ndim, c.numel())
```

```python
print(torch.zeros(5))
# zeros_like(변수): 변수에 들어있는 텐서와 동일한 크기를 가지면서 모든 원소가 0인 새로운 텐서 생성
print(torch.zeros_like(c))
print(torch.arange(3, 10, 2))
print(torch.arange(0, 1, 0.1))
print(torch.linspace(0, 1, 10))
```

#### ◈ 텐서의 연산
: 텐서에 대한 수학 연산, 삼각함수, 비트연산, 집계 등 제공

```python
d = torch.tensor([1, 2, 3])
e = torch.tensor([4, 5, 6])
f = d + e
print(f)
```

```python
g = torch.tensor([[1, 2, 3], [1, 2, 3]])
h = torch.tensor([[4, 5, 6], [1, 1, 1]])
print(g + h)
print(g - h)
print(g * h)
print(g / h)
print(g**2)
```

#### ◈ 행렬곱(dot product)

```python
k = torch.tensor([[1, 2], [3, 4]])
m = torch.tensor([[1, 2], [3, 4]])
print(k @ m)
print(k.matmul(m))
print(torch.mm(k, m))
```
